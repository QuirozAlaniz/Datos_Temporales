### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Escribe tu código aquí
taxis["hora"] = taxis["pickup"].dt.hour
taxis["dia_num"] = taxis["pickup"].dt.dayofweek
taxis["dia_nombre"] = taxis["pickup"].dt.day_name()
taxis["fin_semana"] = (taxis["dia_nombre"]=="Saturday") | (taxis["dia_nombre"]=="Sunday")
taxis["fecha"] = taxis["pickup"].dt.date
taxis["cruza_medianoche"] = taxis["pickup"].dt.day != taxis["dropoff"].dt.day
```

```python
print(f'El número de registros que cruzan la media noche es:\n  {len(taxis[taxis["cruza_medianoche"] == True])}')
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
# Escribe tu código aquí
agrupacion = taxis.groupby("hora").agg(numero_viajes = ("duracion_min","count"), mediana_duracion_min = ("duracion_min", "median")) #Agrupamos por hora y creamos columnas
                                                                                                                                    #con funciones de agregación count y median
```

```python
maximo = agrupacion["numero_viajes"].idxmax()
minimo = agrupacion["numero_viajes"].idxmin()
print(f'La hora con más viajes es: {maximo} con {agrupacion["numero_viajes"].max()} viajes')
print()
print(f'La hora con menos viajes es: {minimo} con {agrupacion["numero_viajes"].min()} viajes')
```

```python
#Código no solicitado pero para responder a la pregunta 3.4
agrupacion2 = taxis.groupby("hora").agg(mediana_duracion_trayecto = ("duracion_min", "median"))
print(agrupacion2.sort_values(by = "mediana_duracion_trayecto", ascending = False).head(3))
```
